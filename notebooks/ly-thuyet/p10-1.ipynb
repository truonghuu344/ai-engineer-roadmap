```python
def kiem_tra_repo(d):
    loi = []
    rd = os.path.join(d, "README.md")
    if not os.path.exists(rd):
        loi.append("thiếu README.md")
    else:
        t = open(rd).read()
        m = re.search(r"^## (Kết quả|Results)\s*\n(.*?)(?=^## |\Z)", t, re.M | re.S)
        if not m or not re.search(r"\d", m.group(2)):
            loi.append("README: mục kết quả chưa có con số nào")
        if not re.search(r"pip install|python ", t):
            loi.append("README: chưa có hướng dẫn chạy")
    if not any(os.path.exists(os.path.join(d, f)) for f in ("requirements.txt", "pyproject.toml")):
        loi.append("thiếu requirements.txt / pyproject.toml")
    if not os.path.isdir(os.path.join(d, "tests")):
        loi.append("thiếu thư mục tests/")
    for root, _, files in os.walk(d):
        for f in files:
            p = os.path.join(root, f)
            if f == ".env":
                loi.append(f"có file bí mật: {p}")
            elif os.path.getsize(p) > 50 * 1024 * 1024:
                loi.append(f"file quá lớn ({os.path.getsize(p) // 2**20} MB): {p}")
    return loi

for l in kiem_tra_repo("repo-demo"):
    print("-", l)
```

## Bài tập 6: Lập lịch mốc từ danh sách việc {ứng dụng}

(Tự làm trước trên trang bài học; các ô dưới có cả lời giải.)

```python
import math
viec = [("Thu thập 200 tài liệu quy chế", 6, False), ("Chunking + index + tìm kiếm", 10, False),
        ("Bộ eval 60 câu hỏi", 8, True), ("Hybrid search + reranker", 12, False), ("Đo và so sánh trên bộ eval", 6, True),
        ("API FastAPI + Docker", 10, False), ("Giao diện demo + deploy", 8, True), ("README, ảnh demo, bài viết", 8, True)]
GIO_TUAN, DU_PHONG = 15, 1.3
# TODO: tuan_xong[tên việc] = tuần (bắt đầu từ 1) mà việc đó xong
```

```python
tong, tuan_xong = 0.0, {}
for ten, gio, moc in viec:
    tong += gio * DU_PHONG
    tuan_xong[ten] = math.ceil(tong / GIO_TUAN)
    print(f"{'★' if moc else ' '} tuần {tuan_xong[ten]:2d}: {ten} (cộng dồn {tong:.1f} giờ)")
```
