## Bài tập 6: Phân tích phễu ứng tuyển theo kênh {ứng dụng}

(Tự làm trước trên trang bài học; các ô dưới có cả lời giải.)

```python
import numpy as np, pandas as pd
rng = np.random.default_rng(0)
TT = ["gửi", "phản hồi", "phỏng vấn", "offer"]
p = {"job board": [0.12, 0.35, 0.2], "giới thiệu": [0.6, 0.6, 0.3], "freelance": [0.25, 0.5, 0.4], "email trực tiếp": [0.18, 0.4, 0.25]}
so_don = {"job board": 60, "giới thiệu": 8, "freelance": 30, "email trực tiếp": 20}
rows = []
for kenh, n in so_don.items():
    for _ in range(n):
        muc = 0
        for xs in p[kenh]:
            if rng.random() < xs:
                muc += 1
            else:
                break
        rows.append({"kenh": kenh, "xa_nhat": TT[muc]})
df = pd.DataFrame(rows)
# TODO: bảng theo kênh: so_don, ti_le_phan_hoi, ti_le_phong_van, ti_le_offer
```

```python
df["muc"] = df["xa_nhat"].map({t: i for i, t in enumerate(TT)})
bang = df.groupby("kenh").agg(so_don=("muc", "size"),
                              ti_le_phan_hoi=("muc", lambda s: (s >= 1).mean()),
                              ti_le_phong_van=("muc", lambda s: (s >= 2).mean()),
                              ti_le_offer=("muc", lambda s: (s >= 3).mean())).sort_values("so_don", ascending=False)
print(bang.round(3))
```

## Bài tập 7: Công cụ báo giá freelance {ứng dụng}

(Tự làm trước trên trang bài học; các ô dưới có cả lời giải.)

```python
import math
hang_muc = [("Phân tích yêu cầu, chuẩn bị dữ liệu", 8), ("Pipeline RAG + bộ eval 40 câu", 20),
            ("API + giao diện chat đơn giản", 12), ("Deploy, hướng dẫn sử dụng", 6), ("2 vòng chỉnh sửa", 6)]
# TODO: gia_gio_toi_thieu, bao_gia
```

```python
def gia_gio_toi_thieu(thu_nhap_thang, gio_tuan, ti_le):
    return thu_nhap_thang / (gio_tuan * 4.33 * ti_le)

def bao_gia(cac_hang_muc, gia_gio, du_phong):
    chi_tiet = [(ten, gio, gio * gia_gio) for ten, gio in cac_hang_muc]
    tong = math.ceil(sum(c for *_, c in chi_tiet) * (1 + du_phong) / 50) * 50
    return tong, chi_tiet

print(f"giá tối thiểu: {gia_gio_toi_thieu(2000, 30, 0.65):.2f} USD/giờ")
tong, ct = bao_gia(hang_muc, 25, 0.25)
for ten, gio, c in ct:
    print(f"  {ten:38s} {gio:3d} giờ  {c:6.0f} USD")
print(f"  {'Tổng (đã gồm 25% dự phòng)':38s} {sum(g for _, g, _ in ct):3d} giờ  {tong:6.0f} USD")
```
