```python
y = np.array([1., 2., 3.])             # nhãn thật, shape (3,)
y_pred = np.array([[1.], [2.], [4.]])  # model trả về shape (3, 1)
(y_pred - y).shape                     # (3, 3)  ← broadcasting thành ma trận 3x3!
((y_pred - y) ** 2).mean()             # 2.33   sai
((y_pred.ravel() - y) ** 2).mean()     # 0.33   đúng
```

```python
z = np.array([-2., 0.5, 3., -1.])
np.where(z > 0, z, 0)     # [0.  0.5 3.  0. ]  ReLU: giữ số dương, còn lại thành 0
np.clip(z, -1, 1)         # [-1.  0.5 1.  -1. ] chặn trong khoảng [-1, 1]
np.cumsum([1, 2, 3, 4])   # [ 1  3  6 10]       tổng tích lũy
z.argmax()                # 2                   chỉ số của phần tử lớn nhất
np.argsort(-z)[:2]        # [2 1]               chỉ số của 2 phần tử lớn nhất (top-k)
```

## Bài tập 4: Khoảng cách từng cặp {code}

(Tự làm trước trên trang bài học; các ô dưới có cả lời giải.)

```python
import time
import numpy as np

def pairwise_dist(A, B):
    return np.sqrt(((A[:, None, :] - B[None, :, :]) ** 2).sum(-1))

def pairwise_loop(A, B):
    D = np.zeros((len(A), len(B)))
    for i in range(len(A)):
        for j in range(len(B)):
            D[i, j] = np.sqrt(((A[i] - B[j]) ** 2).sum())
    return D

A, B = np.random.rand(500, 64), np.random.rand(500, 64)
t = time.time(); D1 = pairwise_dist(A, B); t1 = time.time() - t
t = time.time(); D2 = pairwise_loop(A, B); t2 = time.time() - t
print(np.allclose(D1, D2), f"nhanh hơn {t2 / t1:.0f} lần")
```

## Bài tập 5: Softmax ổn định số {code}

(Tự làm trước trên trang bài học; các ô dưới có cả lời giải.)

```python
import numpy as np

def softmax(Z):
    Z = np.asarray(Z, dtype=float)
    E = np.exp(Z - Z.max(axis=-1, keepdims=True))   # trừ max từng hàng: số lớn nhất thành 0, exp không tràn
    return E / E.sum(axis=-1, keepdims=True)

P = softmax([[1., 2., 3.], [1000., 1001., 1002.]])
print(P.round(3))          # hai hàng giống nhau: chỉ chênh lệch giữa các logit là quan trọng
print(P.sum(axis=1))       # mỗi hàng cộng lại bằng 1
```

## Bài tập 7: Bảng điểm cả lớp {ứng dụng}

(Tự làm trước trên trang bài học; các ô dưới có cả lời giải.)

```python
import numpy as np
mon = ["Toán", "Lý", "Hóa", "Anh"]
ten = ["An", "Bình", "Chi", "Dũng", "Em"]
diem = np.array([[8, 7, 6, 9],
                 [5, 4, 6, 7],
                 [9, 9, 8, 6],
                 [6, 5, 4, 8],
                 [7, 8, 9, 7]], dtype=float)
# TODO: tb_mon, tb_hs, gioi_nhat, z, n_khong_duoi_5
```

```python
tb_mon = diem.mean(axis=0)
tb_hs = diem.mean(axis=1)
gioi_nhat = ten[tb_hs.argmax()]
z = (diem - tb_mon) / diem.std(axis=0)
n_khong_duoi_5 = (diem >= 5).all(axis=1).sum()
print(dict(zip(mon, tb_mon)), dict(zip(ten, tb_hs)), gioi_nhat, n_khong_duoi_5, sep="\n")
print(z.round(2))
```
