```python
def thanh_tien(gia, so_luong, giam_gia=0):
    """Tính tiền một dòng đơn hàng. giam_gia là tỉ lệ, ví dụ 0.1 = giảm 10%."""
    return gia * so_luong * (1 - giam_gia)

print(thanh_tien(1500, 2))                 # 3000
print(thanh_tien(1500, 2, giam_gia=0.1))   # 2700.0
```

```python
def accuracy(du_doan, nhan_that):
    """Tỉ lệ dự đoán đúng."""
    dung = 0
    for d, t in zip(du_doan, nhan_that):
        if d == t:
            dung = dung + 1
    return dung / len(nhan_that)

print(accuracy([1, 0, 1, 1], [1, 1, 1, 0]))   # đúng 2 trên 4: 0.5
```

```python
class Dataset:                          # đúng khung mà torch.utils.data.Dataset yêu cầu
    def __init__(self, rows):           # chạy khi tạo đối tượng: lưu dữ liệu vào self
        self.rows = rows
    def __len__(self):                  # len(ds) gọi hàm này
        return len(self.rows)
    def __getitem__(self, i):           # ds[i] gọi hàm này
        return self.rows[i]

ds = Dataset(["a", "b", "c"])
print(len(ds), ds[1])                   # 3 b
```

```python
class ReviewDataset:
    def __init__(self, data):
        self.items = []
        for cau, sao in data:
            if sao == 3:
                continue
            self.items.append((cau, 1 if sao >= 4 else 0))
    def __len__(self):
        return len(self.items)
    def __getitem__(self, i):
        return self.items[i]
    def dem_nhan(self):
        dem = {}
        for _, nhan in self.items:
            dem[nhan] = dem.get(nhan, 0) + 1
        return dem

ds = ReviewDataset(reviews)
print(len(ds), ds[0], ds[1], ds.dem_nhan())
```

## Bài tập 10: Thống kê điểm thi bằng thư viện có sẵn {ứng dụng}

(Tự làm trước trên trang bài học; các ô dưới có cả lời giải.)

```python
import random
import statistics as st

random.seed(42)
diem = [random.randint(0, 10) for _ in range(40)]
# TODO: in trung bình, trung vị, độ lệch chuẩn; viết ti_le_dat
```

```python
def ti_le_dat(diem, nguong):
    return sum(1 for d in diem if d >= nguong) / len(diem)

print("trung bình:", st.mean(diem))
print("trung vị:", st.median(diem))
print("độ lệch chuẩn:", round(st.stdev(diem), 2))
print("tỉ lệ đạt từ 5:", ti_le_dat(diem, 5))
```
