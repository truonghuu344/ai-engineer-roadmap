## Bài tập 3: One-hot không dùng vòng for {code}

(Tự làm trước trên trang bài học; các ô dưới có cả lời giải.)

```python
y = np.array([2, 0, 1])
one_hot(y, 3)
# [[0, 0, 1],
#  [1, 0, 0],
#  [0, 1, 0]]
```

```python
def one_hot(y, k):
    return np.eye(k)[y]

# cách khác, không tạo ma trận k×k:
def one_hot2(y, k):
    out = np.zeros((len(y), k))
    out[np.arange(len(y)), y] = 1
    return out
```

## Bài tập 4: Khoảng cách từng cặp {code}

(Tự làm trước trên trang bài học; các ô dưới có cả lời giải.)

```python
import time
import numpy as np

def pairwise_dist(A, B):
    return np.sqrt(((A[:, None, :] - B[None, :, :]) ** 2).sum(-1))

def pairwise_loop(A, B):
    D = np.zeros((len(A), len(B)))
    for i in range(len(A)):
        for j in range(len(B)):
            D[i, j] = np.sqrt(((A[i] - B[j]) ** 2).sum())
    return D

A, B = np.random.rand(500, 64), np.random.rand(500, 64)
t = time.time(); D1 = pairwise_dist(A, B); t1 = time.time() - t
t = time.time(); D2 = pairwise_loop(A, B); t2 = time.time() - t
print(np.allclose(D1, D2), f"nhanh hơn {t2 / t1:.0f} lần")
```
